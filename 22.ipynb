```python
def text_adventure_game():
    print("Welcome to the Adventure!")
    print("You wake up in a mysterious forest. There are two paths ahead.")
    choice1 = input("Do you take the LEFT path or the RIGHT path? (left/right): ").strip().lower()

    if choice1 == "left":
        print("\nYou walk down the left path and find a river.")
        print("The water flows swiftly, creating a mesmerizing sound that echoes through the trees.")
        choice2 = input("Do you SWIM across or FOLLOW the riverbank? (swim/follow): ").strip().lower()
        if choice2 == "swim":
            print("\nYou dive in, fighting the current as storm clouds gather overhead.")
            print("The water is colder than expected, and you can feel the power of the river beneath you.")
            choice3 = input("Do you HEAD for the far shore or GRAB a floating log? (head/grab): ").strip().lower()
            if choice3 == "head":
                print("\nYou power through and reach the far shore, exhausted but safe.")
                print("Your muscles ache, but you've made it across. You discover a sealed chest half-buried in the sand.")
                print("You win!")
            elif choice3 == "grab":
                print("\nYou cling to the log and drift downstream toward the sound of rushing water.")
                print("The roaring gets louder, and you realize you're approaching a waterfall!")
                choice4 = input("Do you JUMP off or HOLD on? (jump/hold): ").strip().lower()
                if choice4 == "jump":
                    print("\nYou leap at the last second and land in a calm pool below.")
                    print("The impact knocks the wind out of you, but you're alive and thrilled by your daring escape.")
                    print("A hidden path leads you to a forgotten shrine. You win!")
                elif choice4 == "hold":
                    print("\nYou ride over the falls and wash up on a rocky shore.")
                    print("Bruised but alive, you struggle to catch your breath as the mist settles around you.")
                    print("You find a map etched into the rocks, pointing toward civilization. You win!")
                else:
                    print("\nInvalid choice. The river sweeps you into unknown waters.")
            else:
                print("\nInvalid choice. The current pulls you back to the starting bank.")
                print("Exhausted and defeated, you decide to reconsider your strategy.")
                print("Perhaps there was wisdom in the path not taken. Game Over!")
        elif choice2 == "follow":
            print("\nYou follow the riverbank and hear distant music drifting through the trees.")
            print("The melody is haunting and beautiful, drawing you deeper into the forest.")
            choice3 = input("Do you INVESTIGATE the music or BUILD a raft? (investigate/build): ").strip().lower()
            if choice3 == "investigate":
                print("\nYou find a camp of friendly travelers huddled around a warm fire.")
                print("They welcome you with open arms and offer you food and shelter.")
                choice4 = input("Do you JOIN their fire or ASK for directions? (join/ask): ").strip().lower()
                if choice4 == "join":
                    print("\nThey share stories of distant lands and gift you a silver compass.")
                    print("With their help and guidance, you find the way out of the forest before dawn breaks.")
                    print("You win!")
                elif choice4 == "ask":
                    print("\nThey listen carefully to your plight and point you toward a safe trail through the hills.")
                    print("One traveler even offers to guide you partway. You escape the forest by sunrise.")
                    print("You win!")
                else:
                    print("\nInvalid choice. The travelers move on without you into the darkness.")
            elif choice3 == "build":
                print("\nYou gather logs and tie them together with strong vines from the forest.")
                print("It takes hours of work, but soon you have a sturdy raft ready to launch.")
                choice4 = input("Do you SAIL downstream or CROSS directly? (sail/cross): ").strip().lower()
                if choice4 == "sail":
                    print("\nThe raft glides smoothly downstream into a misty valley filled with ancient ruins.")
                    print("You explore the crumbling structures and uncover a glowing ancient relic.")
                    print("You win!")
                elif choice4 == "cross":
                    print("\nThe raft wobbles precariously, but through careful balance, you reach the far side safely.")
                    print("Your heart pounding, you spot a hidden trail leading away from the river.")
                    print("A hidden trail leads you home. You win!")
                else:
                    print("\nInvalid choice. The raft falls apart and you retreat to safer ground.")
            else:
                print("\nInvalid choice. You linger by the river until nightfall descends upon you.")
        else:
            print("\nInvalid choice. Lost in the forest, you wander endlessly in circles.")
    elif choice1 == "right":
        print("\nYou walk down the right path and encounter a massive sleeping bear.")
        print("Its chest rises and falls gently, and it appears to be in a deep, peaceful slumber.")
        choice2 = input("Do you SNEAK past or RUN away? (sneak/run): ").strip().lower()
        if choice2 == "sneak":
            print("\nYou quietly sneak past the bear, moving with careful, deliberate steps.")
            print("You hold your breath as you pass mere feet from its powerful paws.")
            print("Beyond the bear, you find a hidden cave entrance glowing with an otherworldly light.")
            print("Inside, you discover ancient artifacts of untold value. You win!")
        elif choice2 == "run":
            print("\nYou run away, crashing through the underbrush as fast as your legs can carry you.")
            print("The bear wakes up and chases you with thunderous footsteps shaking the ground.")
            print("You escape, but end up back where you started, breathless and terrified.")
        else:
            print("\nInvalid choice. The bear wakes up suddenly and you must flee for your life!")
    else:
        print("\nInvalid choice. You remain lost in the forest as darkness falls around you.")

# Run the game
text_adventure_game()
```
