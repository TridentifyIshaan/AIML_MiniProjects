```python
import pandas as pd
import matplotlib.pyplot as plt
import statistics as stats
import numpy as np
import seaborn as sns
```

```python
df = pd.read_csv('datasets/student_performance.csv')
df.head()
```

```python
avg_study_hours = np.mean(df['weekly_self_study_hours'])
avg_attendance = np.mean(df['attendance_percentage'])
avg_class_participation = np.mean(df['class_participation'])
avg_total_score = np.mean(df['total_score'])
print(f"Average Weekly Study hours: {avg_study_hours}, Average Attendance: {avg_attendance}, Average Class Participation: {avg_class_participation}, Average Total Score: {avg_total_score}")
```

```python
median_study_hours = np.median(df['weekly_self_study_hours'])
median_attendance = np.median(df['attendance_percentage'])
median_class_participation = np.median(df['class_participation'])
median_total_score = np.median(df['total_score'])
print(f"Median Weekly Study hours: {median_study_hours}, Median Attendance: {median_attendance}, Median Class Participation: {median_class_participation}, Median Total Score: {median_total_score}")
```

```python
mode_study_hours = stats.mode(df['weekly_self_study_hours'])
mode_attendance = stats.mode(df['attendance_percentage'])
mode_class_participation = stats.mode(df['class_participation'])
mode_total_score = stats.mode(df['total_score'])
print(f"Mode Study Hours: {mode_study_hours}, Mode Attendance: {mode_attendance}, Mode Class Participation: {mode_class_participation}, Mode Total Score: {mode_total_score}")
```

```python
sd_study_hours = np.std(df['weekly_self_study_hours'])
sd_attendance = np.std(df['attendance_percentage'])
sd_class_participation = np.std(df['class_participation'])
sd_total_score = np.std(df['total_score'])
print(f"Standard Deviation Study Hours: {sd_study_hours}, Standard Deviation Attendance: {sd_attendance}, Standard Deviation Class Participation: {sd_class_participation}, Standard Deviation Total Score: {sd_total_score}")
```

```python
var_study_hours = np.var(df['weekly_self_study_hours'])
var_attendance = np.var(df['attendance_percentage'])
var_class_participation = np.var(df['class_participation'])
var_total_score = np.var(df['total_score'])
print(f"Variance Study Hours: {var_study_hours}, Variance Attendance: {var_attendance}, Variance Class Participation: {var_class_participation}, Variance Total Score: {var_total_score}")
```

```python
fig, axes = plt.subplots(1, 4, figsize=(20, 5))
sns.set_context("paper", font_scale=1.2)

sns.boxplot(data=df, x='weekly_self_study_hours', color = 'lightblue', ax = axes[0])
sns.boxplot(data=df, x='attendance_percentage', color = 'green', ax = axes[1])
sns.boxplot(data=df, x='class_participation', color = 'orange', ax = axes[2])
sns.boxplot(data=df, x='total_score', color = 'red', ax = axes[3])
```
