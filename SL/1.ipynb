```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
```

```python
df = pd.read_csv('datasets/lifesat/lifesat.csv')
df.head()
```

```python
X = df[['GDP per capita (USD)']].values
y = df['Life satisfaction'].values

# Visualize the data
df.plot(kind='scatter', x='GDP per capita (USD)', y='Life satisfaction')
plt.axis([23_500, 62_500, 4, 9]) # _ is used for better readability of large numbers and this range means x-axis from 23,500 to 62,500 and y-axis from 4 to 9
plt.show()
```

```python
# Select a linear regression model
l = LinearRegression()

# Train the model
l.fit(X, y)

# Make predictions for Cyprus
X_1 = [[37_655.2]] # GDP per capita for Cyprus in 2020
print(l.predict(X_1))
```

```python
# Make predictions for India
X_2 = [[1907]]
print(l.predict(X_2))
```

## KNN
- Here we take average of the neighbours points

```python
from sklearn.neighbors import KNeighborsRegressor
k = KNeighborsRegressor(n_neighbors=3)

k.fit(X, y)

# Make predictions for Cyprus and India
print(k.predict(X_1))
print(k.predict(X_2))
```
