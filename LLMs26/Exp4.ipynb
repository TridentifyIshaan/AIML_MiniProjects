```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
import sacrebleu
import torch

print("Libraries imported.")

# Step 3: Load NLLB model and tokenizer
print("⬇Loading the NLLB model...")
model_name = 'facebook/nllb-200-distilled-600M'
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

# Step 4: Data
source_sentences = [
    "The cat is sitting on the mat.",
    "I love to learn about machine learning.",
    "The weather today is beautiful and sunny."
]

reference_translations = [[
    "बिल्ली चटाई पर बैठी है।",
    "मुझे मशीन लर्निंग के बारे में सीखना पसंद है।",
    "आज का मौसम खूबसूरत और धूप वाला है।"
]]

# Step 5: Translate using model.generate
print("\nTranslating sentences...")
machine_translations = []

# Get the language token ID for Hindi using the correct method
hindi_code_id = tokenizer.convert_tokens_to_ids("hin_Deva")

for text in source_sentences:
    # Prepare the input
    inputs = tokenizer(text, return_tensors="pt")

    # Generate translation using the target language token
    translated_tokens = model.generate(
        **inputs,
        forced_bos_token_id=hindi_code_id,
        max_length=128
    )

    # Decode result
    translation = tokenizer.batch_decode(translated_tokens, skip_special_tokens=True)[0]
    machine_translations.append(translation)

print("Translation complete.")

# Step 6: BLEU
bleu = sacrebleu.corpus_bleu(machine_translations, reference_translations)

# Step 7: Results
print("\n--- Results ---")
for i, (src, res) in enumerate(zip(source_sentences, machine_translations)):
    print(f"\nSource:    {src}")
    print(f"Machine:   {res}")

print(f"\nFinal BLEU Score: {bleu.score:.2f}")
```

```python
# Step 1: Install required libraries
#!pip -q install -U transformers sacrebleu sentencepiece

# Step 2: Import libraries
import torch
import sacrebleu
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

# Step 3: Load the English-to-Hindi translation model
model_name = "Helsinki-NLP/opus-mt-en-hi"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

# Step 4: Define source sentences
source_sentences = [
    "The cat is sitting on the mat.",
    "I love to learn about machine learning.",
    "The weather today is beautiful and sunny."
]

# Step 5: Generate translations
inputs = tokenizer(
    source_sentences,
    return_tensors="pt",
    padding=True,
    truncation=True
)

with torch.no_grad():
    translated_tokens = model.generate(
        **inputs,
        max_length=100
    )

machine_translations = tokenizer.batch_decode(
    translated_tokens,
    skip_special_tokens=True
)

# Step 6: Define human/reference translations
reference_translations = [[
    "बिल्ली चटाई पर बैठी है।",
    "मुझे मशीन लर्निंग के बारे में सीखना पसंद है।",
    "आज का मौसम खूबसूरत और धूप वाला है।"
]]

# Step 7: Calculate BLEU score
bleu = sacrebleu.corpus_bleu(
    machine_translations,
    reference_translations
)

# Step 8: Display translation results
print("--- Translation Results (English to Hindi) ---")

for i in range(len(source_sentences)):
    print(f"Source:    {source_sentences[i]}")
    print(f"Reference: {reference_translations[0][i]}")
    print(f"Machine:   {machine_translations[i]}")
    print()

# Step 9: Display BLEU score
print("--- Assessment ---")
print(f"BLEU Score: {bleu.score:.2f}")
```

```python
# Step 1: Install/upgrade required libraries
#!pip -q install -U transformers sacrebleu sentencepiece

# Step 2: Import libraries
from transformers import pipeline
import sacrebleu

# Step 3: Load the English-to-Hindi translation model
translator = pipeline(
    "translation_en_to_hi",
    model="Helsinki-NLP/opus-mt-en-hi"
)

# Step 4: Define source sentences
source_sentences = [
    "The cat is sitting on the mat.",
    "I love to learn about machine learning.",
    "The weather today is beautiful and sunny."
]

# Step 5: Define human/reference translations
reference_translations = [[
    "बिल्ली चटाई पर बैठी है।",
    "मुझे मशीन लर्निंग के बारे में सीखना पसंद है।",
    "आज का मौसम खूबसूरत और धूप वाला है।"
]]

# Step 6: Generate machine translations
machine_translations_output = translator(source_sentences)

machine_translations = [
    t["translation_text"]
    for t in machine_translations_output
]

# Step 7: Calculate BLEU score
bleu = sacrebleu.corpus_bleu(
    machine_translations,
    reference_translations
)

# Step 8: Display translation results
print("--- Translation Results (English to Hindi) ---")

for i in range(len(source_sentences)):
    print(f"Source:      {source_sentences[i]}")
    print(f"Reference:   {reference_translations[0][i]}")
    print(f"Machine:     {machine_translations[i]}")
    print()

# Step 9: Display assessment
print("--- Assessment ---")
print(f"BLEU Score: {bleu.score:.2f}")
```
