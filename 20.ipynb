## Amazon Bedrock Flows

Amazon Bedrock Flows allow user to easily create a visual workflow that can chain multiple prompts & Bedrock tools, simplifying and accelrating the creation, testing and deployment of LLM-based workflow. Eg- We can write a lambda function that will fetch top 3 news articles, then write a prompt that will summarize the topics that each of them, then use a summary as a variable for the next flow, which can be, for eg., sending an email.

Amazon Bedrock flows helped us to chain lambda functions, knowledge base and LLMs into a workflow. Users can invoke the flow end-to-end. It's just like a pipeline with a drag & drop interface and APIs for linking models, prompts and services into workflows.

# Summary

## Evolution of AI Systems

- Early Generative AI Phase: Used for simple content creation through prompts, producing outputs like paragraphs, images, or code. Characterized as singleton, reactive, and stateless.

- Agent for Generative AI Phase: Leverages reasoning frameworks like ReAct, breaks complex tasks into simple steps, accesses external information, and integrates with APIs.

- Multi-Agent Phase: Systems consisting of multiple fully autonomous agents that seamlessly execute multi-step workflows and adapt to changing environments.

## Agentic AI Fundamentals

- Definition: Autonomous software systems that use AI to reason, plan, and complete tasks on behalf of humans or other systems.

- Core Capabilities: Performs multi-step tasks, maintains context, accesses various data sources, automates workflows, and thinks iteratively.

- Development Drivers: New models (multimodal, domain-specific), tools (orchestration, simulation, debugging), and frameworks with advanced features.

## Building Agentic AI Systems

1. Workflow Agent: For tasks with predetermined stages

 - Serial Pattern (prompt chaining)

 - Parallelization Pattern

 - Routing Pattern

 - Orchestration Pattern

2. Autonomous Agent: Systems that can self-instruct

 - Agent Loop: Perceive → Reason → Act

 - ReAct Architecture (Reason and Action): Integrates step-by-step reasoning with tools

 - RWO Architecture (Reasoning Without Observation): Separates reasoning into modular components

3. Hybrid Agent: Automatically determines when to use workflow or autonomous agent

4. Multi-Agent System: Coordinates multiple specialist agents to solve complex problems

## AWS Support for Agentic AI

### Specialized Solutions

- Amazon Q Developer: Assistant for building application code, automating testing processes, and creating documentation

- Amazon Q Business: Enterprise generative AI assistant integrated with company data sources

- Kiro: Evolution of AI-assisted development tools that orchestrate entire software development projects

### Fully Managed Solutions

- Amazon Bedrock Agent: Enables building autonomous agents with components:

 - Foundation Model

 - Agent Instructions

 - Action Group

 - Knowledge Base

 - Guardrail (security mechanism)

- Amazon Bedrock Agent Core: For moving agent prototypes to production systems with custom runtime features, identity management, observability, etc.

### DIY (Do It Yourself) Solutions

- Strand Agents: Open source SDK for building AI agents with a model-based approach

- A2A Protocol (Agent-to-Agent): Open standard for inter-agent communication

- MCP (Model Context Protocol): Open source standard for connecting AI systems to external tools

## AWS Advantages in Agentic AI

- Robust and secure infrastructure

- Various agent options for each workflow

- Flexibility with Amazon Bedrock

- Support for open source and inter-agent communication protocols
