## Handling both Numerical & Categorical Data

```python
cols = list(df.columns) # to get the column names of the dataset

# 1. Define the threshold (50% of total rows)
threshold = df.shape[0] * 0.5

# 2. Identify which columns to be dropped and which to be kept
cols_to_drop = [i for i in cols if df[i].isnull().sum() > threshold]
cols_to_impute = [i for i in cols if df[i].isnull().sum() <= threshold]

# 3. Imputing values
for i in cols_to_impute:
    if df[i].isnull().any():
        # Categorical
        if df[i].dtype == 'str' or df[i].dtype == 'object':
            df[i] = df[i].fillna(df[i].mode()[0]) # filling the null values with mode of the column
        # Numerical
        else:
            df[i] = df[i].fillna(df[i].median()) # filling the null values with median of the column
```
