## Conditional Indexing

```python
# dictionary
fruit_protein = {
    "Avocado": 2.0,       # grams of protein
    "Guava": 2.6,
    "Blackberries": 2.0,
    "Oranges": 0.9,
    "Banana": 1.1,
    "Apples": 0.3,
    "Kiwi": 1.1,
    "Pomegranate": 1.7,
    "Mango": 0.8,
    "Cherries": 1.0
}
```

```python
# Creating Series via Dictionary
s2 = pd.Series(fruit_protein, name = "Protein")
s2
```

```python
# Conditional Selection:
s2[s2>1]
```

### Logical Operators: and, or, not

```python
s2[(s2>0.5) | (s2 <= 2)]
```

```python
s2[~(s2>1)]
```

## Dataframes

```python
# DataFrame:

data = {
    "Name": ["Alice_Fernandes", "Bob", "Charlie", "David", "Eve", "Alice_Fernandes"],
    "Age": [25, 30, 35, np.nan, 29, 25],
    "Department": ["HR", "IT", "Finance", "IT", "HR", "HR"],
    "Salary": [50000, 60000, 70000, 62000, np.nan, 50000]
}
data
```

## Data Cleaning

```python
df.isnull().sum()
```

```python
df.dropna(how = "any") # any row that had any null value
```

```python
df.dropna(how = "all") #if all the values in any row are null then we drop that row
```

```python
df
```

```python
df["Age"].fillna(df["Age"].mean())
```

```python
df["Salary"].fillna(df["Salary"].median())
```

```python
df["Age"].ffill(inplace=True) # forward fill
```

```python
df["Age"].bfill(inplace=True) #backward fill
```

```python
df["Name"] = df["Name"].replace("Charlie", "Rose")

df
```

## Duplicates

```python
# duplicates
df_dup = df[df.duplicated(keep = "last")]
df_dup
```

```python
df = df.drop_duplicates()
df
```

## Lambda Functions

```python
# Lambda -> python

df["Promoted Salary"] = df["Promoted Salary"].apply(lambda x: x/10 if x > 650000 else x)
df
```

```python
name = "Alice_Fernandes"
df[["first_name", "last_name"]] = df["Name"].str.split("_", expand = True) # expand will set the value after _ to None for those who don't have _
df
```

```python
# apply and lambda
def multiplying_age(x):
  return x*2

df["Age"] = df["Age"].apply(multiplying_age)
df
```

```python
df["Age"] = df["Age"].apply(lambda x: x/2)
df
```

## Joins & Merges

```python
department_info = {
    "Dept": ["HR", "IT", "Finance"],
    "Location": ["New York", "San Francisco", "Chicago"],
    "Manager": ["Laura", "Steve", "Nina"]
}

df2 = pd.DataFrame(department_info)
df2
```

```python
pd.concat([df, df2])
```

```python
pd.concat([df, df2], axis = 1) # Common column is repeated
```

```python
pd.merge(df, df2, on = "Dept") # Merged on the common column Dept
```
