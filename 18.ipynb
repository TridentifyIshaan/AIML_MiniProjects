# Introduction

Given a square binary matrix mat[][] representing a maze. A rat starts at the top-left corner (0,0) and needs to reach the bottom-right corner (n-1, n-1). The rat can move in four directions: Up (U), Down (D), Left (L), Right (R). Find all possible paths from (0, 0) to (n-1, n-1). If multiple paths exist, return them in lexicographically sorted order otherwise If no path exists, return empty list.

## Learning

How search algorithms like Depth-First Search (DFS) can be used to solve it.

## Conditions

A rat cannot visit the same cell more than once in a path.
1 represents an open cell (rat can visit), and 0 represents a blocked cell (rat cannot visit).

## Example

**Input:** _mat[][] = [[1, 0, 0, 0], [1, 1, 0, 1], [1, 1, 0, 0], [0, 1, 1, 1]]_

## Solution

**Output:** _["DDRDRR", "DRDDRR"]_
**Explanation:** _The possible paths are: DDRDRR, DRDDRR_

## Function to perform DFS to solve the rat in a maze problem

### Mapping

Directions: Down, Left, Right, Up
`dir = "DLRU"` <br>
`dr = [1, 0, 0, -1]` <br>
`dc = [0, -1, 1, 0]` <br>

`D` → `(dr=1, dc=0)` move down one row <br>
`L` → `(dr=0, dc=-1)` move left one column <br>
`R` → `(dr=0, dc=1)` move right one column <br>
`U` → `(dr=-1, dc=0)` move up one row <br>

```python
# Directions: Down, Left, Right, Up
dir = "DLRU"
dr = [1, 0, 0, -1] #row
dc = [0, -1, 1, 0] #column
```

### Checking if the rat stay inside the maze

- Ensures 0 ≤ r < n and 0 ≤ c < n so positions stay inside the `n × n` maze.

```python
def isValid(r, c, n, maze):
    return r >= 0 and c >= 0 and r < n and c < n and maze[r][c] == 1
```

### Function to find all valid paths

```python
def findPath(r, c, maze, path, res):
    n = len(maze)

    # If destination is reached, store the path
    if r == n - 1 and c == n - 1:
        res.append("".join(path))
        return
    
    # Mark current cell as visited
    maze[r][c] = 0

    for i in range(4):
        nr, nc = r + dr[i], c + dc[i]
        if isValid(nr, nc, n, maze):
            path.append(dir[i])
            
            # Move to the next cell recursively
            findPath(nr, nc, maze, path, res)
            
            # Backtrack
            path.pop()
    
    # Unmark current cell
    maze[r][c] = 1
```

### Function to find all paths and return them

```python
def ratInMaze(maze):
    result = []
    n = len(maze)
    path = []

    if maze[0][0] == 1 and maze[n - 1][n - 1] == 1:
        
        # Start from (0,0)
        findPath(0, 0, maze, path, result)

    # Sort results lexicographically
    result.sort()
        
    return result
```

### Main

```python
if __name__ == "__main__":
    maze = [
        [1, 0, 0, 0],
        [1, 1, 0, 1],
        [1, 1, 0, 0],
        [0, 1, 1, 1]
    ]

    result = ratInMaze(maze)

    for p in result:
        print(p, end=" ")
```

**Time Complexity:** O(4n*n), because on every cell we have to try 4 different directions. <br>
**Auxiliary Space:** O(n2), Maximum Depth of the recursion tree.
