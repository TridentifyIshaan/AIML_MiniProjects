````python
df_country=pd.read_excel('Country-Code.xlsx')
df_country.head()
````

````python
final_df = pd.merge(df, df_country, on='Country Code', how='left')
````

````python
final_df.head(2)
````

````python
## To Check Data Types
final_df.dtypes
````

````python
final_df.columns
````

````python
final_df.Country.value_counts()
````

````python
country_names=final_df.Country.value_counts().index
````

````python
country_val=final_df.Country.value_counts().values
````

````python
## Pie Chart - Top3 countries that uses Zomato
plt.pie(country_val[:3], labels=country_names[:3], autopct='%1.2f%%')
````

````python
## Relationships

final_df.columns
````

````python
ratings=final_df.groupby(['Aggregate rating', 'Rating color', 'Rating text']).size().reset_index().rename(columns={0:'Rating Count'})
````

````python
ratings
````

````python
ratings.head()
````

````python
plt.rcParams['figure.figsize']=(12,6)
sns.barplot(x='Aggregate rating', y='Rating Count', data=ratings, hue='Rating color', palette=['beige', 'red', 'orange', 'yellow', 'green', 'olive'])
````

````python
## Count plot
sns.countplot(x='Rating color', data=ratings, palette=['beige', 'red', 'orange', 'yellow', 'green', 'olive'])
````

````python
## Find the countries name that has given 0 rating
final_df[final_df['Rating color']=='White'].groupby('Country').size().reset_index().head(5)
````

````python
## Find out which currency is used by which country?
final_df[['Country', 'Currency']].groupby(['Country', 'Currency']).size().reset_index()
````

````python
## Which countries do have online deliveries option?

final_df[final_df['Has Online delivery']=='Yes'].Country.value_counts()
````
