## Contexts

- set_context() allows us to override default parameters. This affects things like the size of the labels, lines, and other elements of the plot, but not the overall style -> set_style(). The context are:
    - poster - Very Large elements, suitable for large displays or posters
    - Paper - Ideal for printing in academic papers.
    - Notebook - Good for working inside Jupyter notebooks (default setting).
    - Talk - Good for presentation slides where readability matters.

```python
fig, axes = plt.subplots(1, 3, figsize=(20, 5))
sns.set_context("paper")

sns.scatterplot(data = penguins, x = "flipper_length_mm", y = "body_mass_g", hue = "island", ax = axes[0])
axes[0].set_title("By Island")
sns.scatterplot(data = penguins, x = "flipper_length_mm", y = "body_mass_g", hue = "sex", ax = axes[1])
axes[1].set_title("By Sex")
sns.scatterplot(data = penguins, x = "flipper_length_mm", y = "body_mass_g", hue = "species", ax = axes[2])
axes[2].set_title("By Species")
```

## Palettes

- Set 1 - Bright & high-contrast colors (Distinct Colors = 9)
- Set 2 - Softer, more pastel-like colors (Distinct Colors = 8)
- Set 3 - A mix of soft & bold, varied hues (Distinct Colors = 12)
- deep - Default pallete, rich and balanced
- muted - softer colors, good for many categories
- bright - High-contrast, bold colors
- pastel - Soft, pale tones
- dark - Darker tones, easy to distinguish
- colorblind - Optimized for colorblind accessibility



```python
sns.set_context("poster")

fig, axes = plt.subplots(5, 5, figsize=(30, 35))
axes_flat = axes.flatten() # So we can use loops easily

all_palettes = [
    "Set1", "Set2", "Set3", "deep", "muted",
    "bright", "pastel", "dark", "colorblind", "Pastel1",
    "Dark2", "Accent", "Blues", "Greens", "Purples",
    "Greys", "OrRd", "YlGnBu", "BuPu", "coolwarm",
    "RdBu", "BrBG", "PiYG", "Spectral", "terrain_r"
]

# Loop through all palettes
for i, pal in enumerate(all_palettes): #Enumerate - Loop through a list to keep track of index as well as value
    ax = axes_flat[i]
    
    sns.scatterplot(data = penguins, x="flipper_length_mm",  y="body_mass_g",  hue="species", palette=pal, ax=ax)
    
    # Update title so you know which palette is which
    ax.set_title(f"Species (Palette: {pal})")

plt.tight_layout()
plt.show()
```

## Heatmap

Heatmap: a 2D color-coded matrix that shows relationships or patterns between 2 variables.
It's commonly used to:
- Show correlation between features
- Visualize confusion matrices -> will see in the later stages.
- Highlight missing data -> sns.heatmap(df.isnull())

```python
columns =["bill_length_mm", "bill_depth_mm", "flipper_length_mm", "body_mass_g"]
penguins [columns].corr()
```

```python
sns.heatmap(data = penguins [columns].corr(), annot = True)
```

```python
sns.heatmap(data = penguins [columns].corr(), annot = True, vmin = -2, linewidth = "2", linecolor="black")
```

## Pair Grid

```python
# pair grid -> subplot grid for plotting pairwise relationships in a dataset.
graph = sns.PairGrid(data = penguins, hue = "sex", palette = "Set2")
graph.map_upper(sns.scatterplot) # upper triangle
graph.map_lower(sns.kdeplot) # lower triangle
graph.map_diag(sns.histplot) # diagonal
graph.add_legend()
```
